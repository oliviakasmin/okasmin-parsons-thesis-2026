```python
from pathlib import Path
import pandas as pd

csv_path = Path("object_function_groups.csv")
df = pd.read_csv(csv_path)
```

```python
df.head()
```

```python
final_group_summary = (
    df["final_group"]
    .astype("string")
    .str.strip()
    .value_counts(dropna=False)
    .rename_axis("final_group")
    .reset_index(name="count")
)

final_group_summary["percent"] = (
    final_group_summary["count"] / len(df) * 100
).round(2)

final_group_summary
```

```python
nan_rows = df.loc[df["final_group"].isna()].reset_index(drop=True)

print("NaN final_group rows:", len(nan_rows))
nan_rows
```
