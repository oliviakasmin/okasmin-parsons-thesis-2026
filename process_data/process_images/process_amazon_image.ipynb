```python
import sys
from pathlib import Path

REPO_ROOT = Path("/Users/oliviakasmin/Desktop/okasmin-parsons-thesis-2026")  # or: Path.cwd() if cwd is repo root
PROC_IMG = REPO_ROOT / "process_data" / "process_images"
if str(PROC_IMG) not in sys.path:
    sys.path.insert(0, str(PROC_IMG))

from PIL import Image

from crop_standardize import standardize_single_image
from extract_mask_contours import (
    draw_contours_from_mask_image,
    extract_outline_from_mask_image,
)
from remove_background import run_remove_background_step
from utils import save_images
```

```python
input_path = REPO_ROOT / "app/public/amazon_vase.jpg"
output_dir = REPO_ROOT / "app/public"
object_id = "amazon_vase"  # -> amazon_vase_no_bg.png, amazon_vase_mask.png, amazon_vase_outline.png

img = Image.open(input_path).convert("RGB")
no_bg_image, model_mask = run_remove_background_step(image=img)
standardized, std_mask = standardize_single_image(no_bg_image, model_mask)
contours, _, _, _ = extract_outline_from_mask_image(std_mask)
contours_image = draw_contours_from_mask_image(mask_image=std_mask, contours=contours)

paths = save_images(standardized, std_mask, contours_image, object_id, output_dir=output_dir)
paths
```
